```python
import pandas as pd

health_data = {
    'HighBP':      [1, 0, 1, 0, 1, 1, 0, 0, 1, 0] * 50,
    'Smoker':      [1, 1, 0, 0, 1, 0, 1, 0, 1, 1] * 50,
    'HeartDisease': [1, 0, 0, 0, 1, 1, 0, 0, 1, 0] * 50
}

df = pd.DataFrame(health_data)
df.to_csv('health_dataset.csv', index=False)

display(df.head())
```

```python
df = pd.read_csv('health_dataset.csv')

def estimate_likelihood(high_bp, smoker):
    subset = df[(df['HighBP'] == high_bp) & (df['Smoker'] == smoker)]

    prob_disease = len(subset[subset['HeartDisease'] == 1]) / len(subset) if len(subset) > 0 else 0
    prob_no_disease = 1 - prob_disease

    return {'HeartDisease': prob_disease, 'NoHeartDisease': prob_no_disease}

probabilities = estimate_likelihood(high_bp=1, smoker=1)
print("Conditional Probabilities given HighBP=1 and Smoker=1:")
print(probabilities)
```

```python
print(f"Estimated Likelihood of Heart Disease: {probabilities['HeartDisease'] * 100:.2f}%")
```
